```python
# 第二部分：数据处理函数
def load_data(file_path):
    """加载JSON格式的数据集"""
    with open(file_path, 'r', encoding='utf-8') as f: #change the path
        return json.load(f)

def format_data_for_training(examples):
    """将聊天格式的数据转换为训练格式"""
    formatted_examples = []
    for example in examples:
        messages = example["messages"]
        formatted_text = ""
        for message in messages:
            role = message["role"]
            content = message["content"]
            formatted_text += f"<|{role}|>\n{content}</s>\n"
        formatted_examples.append({"text": formatted_text})
    return formatted_examples

def create_datasets(train_file, val_file, test_file=None, tokenizer=None, max_length=2048):
    """创建训练、验证和测试数据集"""
    train_data = load_data(train_file)
    val_data = load_data(val_file)

    train_formatted = format_data_for_training(train_data)
    val_formatted = format_data_for_training(val_data)

    # 创建原始数据集
    train_dataset_raw = Dataset.from_dict({"text": [item["text"] for item in train_formatted]})
    val_dataset_raw = Dataset.from_dict({"text": [item["text"] for item in val_formatted]})

    # 对数据集进行预处理
    def preprocess_function(examples):
        return tokenizer(
            examples["text"],
            truncation=True,
            max_length=max_length,
            padding="max_length",
        )

    # 进行tokenize处理
    if tokenizer:
        train_dataset = train_dataset_raw.map(preprocess_function, batched=True)
        val_dataset = val_dataset_raw.map(preprocess_function, batched=True)
    else:
        # 如果没有tokenizer，返回原始数据集
        train_dataset = train_dataset_raw
        val_dataset = val_dataset_raw

    test_dataset = None
    if test_file:
        test_data = load_data(test_file)
        test_formatted = format_data_for_training(test_data)
        test_dataset_raw = Dataset.from_dict({"text": [item["text"] for item in test_formatted]})
        if tokenizer:
            test_dataset = test_dataset_raw.map(preprocess_function, batched=True)
        else:
            test_dataset = test_dataset_raw

    return train_dataset, val_dataset, test_dataset
```

```python
# 第三部分：模型准备和训练相关函数
def prepare_model(model_name, bits, lora_r, lora_alpha, lora_dropout):
    """准备模型和tokenizer"""
    print(f"加载模型 {model_name}...")

    # 加载tokenizer
    tokenizer = AutoTokenizer.from_pretrained(
        model_name,
        cache_dir="model_cache",
        trust_remote_code=True
    )
    tokenizer.pad_token = tokenizer.eos_token

    # 加载模型 - 修复参数冲突
    if bits == 4:
        # 4位量化方式
        model = AutoModelForCausalLM.from_pretrained(
            model_name,
            cache_dir="model_cache",
            trust_remote_code=True,
            torch_dtype=torch.bfloat16,
            device_map="auto",
            quantization_config={
                "load_in_4bit": True,
                "bnb_4bit_compute_dtype": torch.bfloat16,
                "bnb_4bit_use_double_quant": True,
                "bnb_4bit_quant_type": "nf4"
            }
        )
    elif bits == 8:
        # 8位量化方式
        model = AutoModelForCausalLM.from_pretrained(
            model_name,
            cache_dir="model_cache",
            trust_remote_code=True,
            torch_dtype=torch.bfloat16,
            device_map="auto",
            load_in_8bit=True
        )
    else:
        # 不进行量化
        model = AutoModelForCausalLM.from_pretrained(
            model_name,
            cache_dir="model_cache",
            trust_remote_code=True,
            torch_dtype=torch.bfloat16,
            device_map="auto"
        )

    # 准备LoRA配置 - 使用正确的参数
    model = prepare_model_for_kbit_training(model)
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"]

    # 使用当前版本支持的参数
    lora_config = LoraConfig(
        r=lora_r,                   # LoRA的秩
        lora_alpha=lora_alpha,      # LoRA的alpha参数
        lora_dropout=lora_dropout,  # LoRA的dropout率
        bias="none",                # 不对偏置应用LoRA
        task_type="CAUSAL_LM",      # 任务类型：因果语言模型
        target_modules = [
          "q_proj",  # Query projection
          "k_proj",  # Key projection
          "v_proj",  # Value projection
          "o_proj",  # Output projection
          "gate_proj", "up_proj", "down_proj"  # FFN部分（LLaMA结构特有）
        ],  # 目标模块列表
        init_lora_weights="gaussian" # 使用高斯初始化方法对生物医学领域有提升
        # 例如：
        # fan_in_fan_out=False,
        # modules_to_save=None,
        # init_lora_weights=True
    )

    model = get_peft_model(model, lora_config)
    model.print_trainable_parameters()

    return model, tokenizer
```

```python
# 第四部分：自定义Trainer类
# 修改CustomTrainer类以监控关键指标
class CustomTrainer(Trainer):
    def compute_loss(self, model, inputs, return_outputs=False, num_items_in_batch=None):
        """计算训练损失和相关指标"""
        outputs = model(**inputs)
        loss = outputs.loss

        # 记录训练损失 - 这是最基础的监控指标
        if self.args.logging_steps > 0 and self.state.global_step % self.args.logging_steps == 0:
            metrics_to_log = {
                "train/loss": loss.item(),
                "train/perplexity": torch.exp(loss).item(),
            }

            # 对分类任务添加accuracy
            if hasattr(outputs, "logits") and "labels" in inputs:
                predictions = outputs.logits.argmax(-1)
                labels = inputs["labels"]
                mask = labels != -100
                if mask.sum() > 0:  # 确保有有效标签
                    accuracy = (predictions[mask] == labels[mask]).float().mean()
                    metrics_to_log["train/accuracy"] = accuracy.item()

            wandb.log(metrics_to_log, step=self.state.global_step)

        return (loss, outputs) if return_outputs else loss

    def evaluate(self, eval_dataset=None, ignore_keys=None, metric_key_prefix="eval"):
        """详细评估函数"""
        # 运行原始评估
        metrics = super().evaluate(eval_dataset, ignore_keys=ignore_keys, metric_key_prefix=metric_key_prefix)

        # 计算并记录基础指标 - validation loss是核心监控指标
        eval_loss = metrics[f"{metric_key_prefix}_loss"]
        perplexity = np.exp(eval_loss)
        metrics[f"{metric_key_prefix}_perplexity"] = perplexity

        # 记录到wandb的指标
        log_metrics = {
            f"{metric_key_prefix}/loss": eval_loss,
            f"{metric_key_prefix}/perplexity": perplexity,
        }

        # 如果是测试集评估，添加更复杂的指标
        if metric_key_prefix == "test" or (hasattr(self, "compute_detailed_metrics") and self.compute_detailed_metrics):
            # 采样部分数据进行详细评估
            sample_size = min(50, len(eval_dataset))
            eval_samples = eval_dataset.select(range(sample_size))

            # 初始化指标计算器
            try:
                from rouge_score import rouge_scorer
                rouge = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)
                has_rouge = True
            except ImportError:
                has_rouge = False
                print("Warning: rouge_score not installed, skipping ROUGE metrics")

            # 生成回答并计算指标
            all_rouge_scores = {"rouge1": [], "rouge2": [], "rougeL": []}
            response_lengths = []

            for sample in eval_samples:
                # 准备输入
                prompt = sample["text"]
                assistant_start = "<|assistant|>\n"

                # 找到最后一个<|assistant|>的位置作为参考答案
                reference_start = prompt.rfind(assistant_start)
                if reference_start == -1:
                    continue  # 跳过没有助手回复的样本

                reference_text = prompt[reference_start + len(assistant_start):].strip()
                user_prompt = prompt[:reference_start].strip()

                # 生成回答
                inputs = self.tokenizer(user_prompt, return_tensors="pt", truncation=True)
                inputs = {k: v.to(self.model.device) for k, v in inputs.items()}

                with torch.no_grad():
                    outputs = self.model.generate(
                        **inputs,
                        max_length=512,
                        do_sample=True,
                        temperature=0.7,
                        top_p=0.9,
                    )

                generated_text = self.tokenizer.decode(outputs[0], skip_special_tokens=True)
                generated_text = generated_text[len(user_prompt):].strip()  # 只保留生成的部分
                response_lengths.append(len(generated_text.split()))

                # 计算ROUGE分数
                if has_rouge and reference_text and generated_text:
                    try:
                        scores = rouge.score(reference_text, generated_text)
                        for metric, score in scores.items():
                            all_rouge_scores[metric].append(score.fmeasure)
                    except Exception as e:
                        print(f"Error computing ROUGE: {e}")

            # 添加额外指标到metrics和wandb
            if has_rouge:
                for metric, scores in all_rouge_scores.items():
                    if scores:  # 确保有分数
                        metrics[f"{metric_key_prefix}_{metric}"] = np.mean(scores)
                        log_metrics[f"{metric_key_prefix}/{metric}"] = np.mean(scores)

            if response_lengths:
                metrics[f"{metric_key_prefix}_response_length"] = np.mean(response_lengths)
                log_metrics[f"{metric_key_prefix}/response_length"] = np.mean(response_lengths)

            # 添加一个CoT路径指标的占位符 - 这需要针对具体任务定制
            if hasattr(self, "eval_cot_accuracy"):
                cot_acc = self.eval_cot_accuracy(eval_samples)
                metrics[f"{metric_key_prefix}_cot_accuracy"] = cot_acc
                log_metrics[f"{metric_key_prefix}/cot_accuracy"] = cot_acc

        # 记录到wandb
        if self.args.report_to == ["wandb"]:
            wandb.log(log_metrics)

        return metrics
```

```python
# 第五部分：超参数优化
def objective(trial, config, train_dataset, val_dataset, tokenizer):
    """Optuna的目标函数，用于超参数优化"""
    # 定义超参数搜索空间
    lr = trial.suggest_float("learning_rate", 1e-5, 5e-4, log=True)
    lora_r = trial.suggest_categorical("lora_r", [64, 128, 192])
    lora_alpha = trial.suggest_categorical("lora_alpha", [128, 256, 384])
    lora_dropout = trial.suggest_float("lora_dropout", 0.01, 0.1)

    # 准备模型
    model, _ = prepare_model(
        config["model_name"],
        config["bits"],
        lora_r,
        lora_alpha,
        lora_dropout
    )

    # 设置训练参数
    training_args = TrainingArguments(
        output_dir=f"{config['output_dir']}_trial_{trial.number}",
        num_train_epochs=1,
        per_device_train_batch_size=config["per_device_train_batch_size"],
        per_device_eval_batch_size=config["per_device_eval_batch_size"],
        gradient_accumulation_steps=config["gradient_accumulation_steps"],
        learning_rate=lr,
        weight_decay=config["weight_decay"],
        warmup_steps=config["warmup_steps"],
        eval_strategy="epoch",
        save_strategy="epoch",
        load_best_model_at_end=True,
        metric_for_best_model="eval_perplexity",
        greater_is_better=False,
        fp16=True,
        optim="adamw_bnb_8bit",
        remove_unused_columns=True,
    )

    # 创建trainer并训练
    trainer = CustomTrainer(
        model=model,
        args=training_args,
        train_dataset=train_dataset,
        eval_dataset=val_dataset,
        data_collator=DataCollatorForLanguageModeling(tokenizer=tokenizer, mlm=False),
        tokenizer=tokenizer,
    )

    # 训练并返回验证集结果
    trainer.train()
    eval_results = trainer.evaluate()

    return eval_results["eval_perplexity"]
```
